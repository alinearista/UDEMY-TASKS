```python
#Challenge: Your Monthly Savings Plan According to Your Age with Interest Rate
#In this Python programming challenge, you will create a personal monthly savings plan based on a user's age,
#taking into account the interest rate on their invested savings. By using Python operators, you will determine
#the ideal savings amount according to the user's age, financial goals, and interest rate.
#This challenge will showcase the practical applications of Python operators in the realm of personal finance.

#Challenge: Your goal is to create a program that calculates the ideal monthly savings amount based on the
#user's age, target retirement age, desired savings amount at retirement, and an annual interest rate.
#To accomplish this, you will use Python operators and the following steps:
#Prompt the user for their current age.
currant_age=int(input('Waht is your age?'))
#Prompt the user for their target retirement age.
target_age=int(input('What age do you plan to retaire?'))
#Prompt the user for their desired savings amount at retirement.
desired_savings=int(input('How much do you want to save?'))
#Prompt the user for the annual interest rate on their savings.
interest_rate=float(input('What is the anual interest rate? (e.g.0.04 for 4%)'))
#Calculate the number of years le until retirement.
years_until_retirment = target_age-currant_age
print(f"You have {years_until_retirment} years until retairment")
#Calculate the total number of months le until retirement.
months_until_retirment = years_until_retirment*12
#calculate the monthly interestrate
monthly_interest_rate=interest_rate/12
print(f"You have {months_until_retirment} months until retairment")
#Determine the ideal monthly savings amount based on the desired savings, the number of months le until retirement, and the interest rate.
monthly_savings=(desired_savings*interest_rate)/((1+monthly_interest_rate)**months_until_retirment-1)
#Display the ideal monthly savings amount to the user.
print(f'Your ideal monthly savings amount is: {monthly_savings:.2f}')
#Discover the capabilities of Python operators in personal finance! Good luck!

```
