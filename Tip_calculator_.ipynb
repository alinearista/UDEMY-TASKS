```python
#Prompt the user for the total cost of the meal.
total_cost=float(input('What is the total cost of the meal?'))
#Prompt the user for the tip percen
tip_percentage=float(input('what percentage would you like to tip?)(e.g. 0.04 for 4%)'))
#Prompt the user for the number or people at the table
number_of_people=int(input('How many people are at the table?'))
#Calcualate the tip amount using the tip percentage and th emeal cost
tip_amount= total_cost * tip_percentage
#calculate the totla cost of the meal, including the tip
total_cost_with_tip= total_cost + tip_amount
#divide the total cost by the number of people at the table
cost_per_person = total_cost_with_tip/number_of_people
#display the tip amount, total cost, and cost perperson to the user
print(f'The tip amount is ${tip_amount:.2f}')
print(f'The total cost of the meal is ${total_cost_with_tip:.2f}')
print(f'The cost per person is ${cost_per_person:.2f}')
```
